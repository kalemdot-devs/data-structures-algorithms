```python
from urllib.parse import urlparse

def get_short_domain(url):
    parsed_url = urlparse(url)
    domain = parsed_url.netloc

    # Remove 'www.' if present
    if domain.startswith("www."):
        domain = domain[4:]

    # Keep only the main domain (example.com)
    parts = domain.split(".")
    if len(parts) > 2:
        domain = ".".join(parts[-2:])

    return domain


# Paste your URL here
url = input("Enter URL: ")
print("Short domain:", get_short_domain(url))
```

```python
urls = ['https://www.google.com/search?q=adriaanbetje&ie=UTF-8', 'https://www.fiverr.com/inbox/adriaanbetje', 'https://www.geeksforgeeks.org/python/python-programming-language-tutorial/', 'https://www.citymd.com']
import csv
for url in urls:
    short_url = url.split('//')
    if 'w.' in short_url[-1]:
        part_1 = url.split('.')[-2]
        part_2 = url.split('.')[-1].split('/')[0]
        domain = f'{part_1}.{part_2}'
        with open('domainx.csv', 'a', encoding='utf-8', newline='') as f:
            csv.writer(f).writerow([domain])
        print(domain)
```
