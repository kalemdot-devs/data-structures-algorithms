### Check num is Plindrome

```python
num = int(input('Enter your number : '))
```

```python
revese = 0
if num < 0:
    print('False')

while num != 0:
    last_digit = num % 10 
    revese = revese + (last_digit * 10)
    num = num // 10

print(revese)
```

#### Hollow Rectangle

```python
rows = 4
cols = 5

for i in range(1, rows+1):
    for j in range(1, cols+1):

        if i == 1 | i == rows | j == 1 | j==cols:
            print("*", end='')
        else:
            print(" ", end='')
    print()
```
